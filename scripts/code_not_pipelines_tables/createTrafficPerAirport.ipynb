# API Experiment

Since the dataset is so big, the ISTAC doesnt give you the option of downloading just one with all the aircraft movements (arrival, departure, all) and all the types of observed values (passenger, mail & goods, operations). For each type of aircraft movement and for each type of obs_val you have to make an API call, so if you wanted to get all the data you would need 3x3 = 9 API calls, each one retrieving essentially the same data except for the obs_value column. And the call is not fast either, trying it in the `data_exploration` notebook it lasted one minute and twenty seconds for one call

Since this code is to create the table for the first time I dont mind doing the 9 API calls, however in the pipeline...

Try the API call bringing only the last month to see how long it last, if its fast then I dont mind retreiving the three `Total` tables (where aircraft movement is total). But if it is slow then only retrieve `Arrival` and `Departure`

Unlike with TrafficPerTerritory, the datasets for each observed value do not have the same destination airports, so we will have to do a full outer join

```python
print(op_t.shape[0] == pass_t.shape[0])
print(op_t.shape[0])
print(pass_t.shape[0])
```

```python
set_pass = set(pass_t['AEROPUERTO_ESCALA_CODE'].unique())
set_op = set(op_t['AEROPUERTO_ESCALA_CODE'].unique())

only_in_A = set_pass - set_op
only_in_B = set_op - set_pass
in_both = set_pass & set_op
```

```python
print("Only in passenger:", only_in_A)
print("Only in operations:", only_in_B)
print("In both:", in_both)
```

First fill na with 0 in the dataframes we are going to merge, since we are doing an outer join and the result will contain nan values, we dont want to confuse nan because is 0 and nan because the observed value doesnt belong to the destination airport. In theory it doesnt matter because in the end we will fill all with 0 anyways, but we do it at first to check if the union is correct

```python
pass_t['OBS_VALUE'] = pass_t['OBS_VALUE'].fillna(0)
op_t['OBS_VALUE'] = op_t['OBS_VALUE'].fillna(0)
```

```python
pass_t
```

```python
df_f = pass_t.merge(op_t, 
                    on=['AEROPUERTO_BASE_CODE', 'AEROPUERTO_ESCALA_CODE', 'MOVIMIENTO_AERONAVE_CODE', 'SERVICIO_AEREO_CODE', 'TIME_PERIOD_CODE'], 
                    suffixes=("", "_op"),
                    how="outer")
df_f.drop(columns=[col for col in df_f.columns if col.endswith('op') and not col in ['OBS_VALUE_op']], inplace=True)
```

Because we filled nan with 0 previous to the join, there shouldnt be any rows where both observed values are null

```python

df_f.loc[pd.isna(df_f['OBS_VALUE_op']) & pd.isna(df_f['OBS_VALUE'])]
```

```python
# We dont need this
df_f.drop('MEDIDAS_CODE', axis=1, inplace=True)
```

The null values in operations should be way higher than in passengers, because passengers has that much more airports

```python
print(f"Number of null observed values in operation after join: {len(df_f.loc[pd.isna(df_f['OBS_VALUE_op'])])}")
print(f"Number of null observed values in passengers after join: {len(df_f.loc[pd.isna(df_f['OBS_VALUE'])])}")
```

In fact, to check if everything is correct, the number of null in OBS_VALUE in the join should be:

In passengers: (number of airports that passenger doesnt have compared to operations) * (len unique rest of columns used in month)\
In operations: (number of airports that operations doesnt have compared to passengers) * (len unique rest of columns used in month)

And the total number of rows in the join dataset should be:
(number of rows of the dataset with more airports: passenger) + (number of null OBS_VALUE in passenger)

```python
pass_t.shape[0] + len(df_f.loc[pd.isna(df_f['OBS_VALUE'])]) == df_f.shape[0]
```

```python
territory = pd.read_csv('../../data/Final_Territory.csv')
airservice = pd.read_csv('../../data/AirService.csv')
aircraftmovement = pd.read_csv('../../data/Final_AircraftMovement.csv')
airport = pd.read_csv('../../data/Airport.csv')
```

```python
# Replace values in df_f["TERRITORIO_CODE"] with corresponding TerritoryId
df_f["AEROPUERTO_BASE_CODE"] = df_f["AEROPUERTO_BASE_CODE"].map(dict(zip(airport["AirportCode"], airport["AirportId"])))

# Replace values in df_f["AEROPUERTO_ESCALA_CODE"] with corresponding TerritoryId
df_f["AEROPUERTO_ESCALA_CODE"] = df_f["AEROPUERTO_ESCALA_CODE"].map(dict(zip(airport["AirportCode"], airport["AirportId"])))
# Delete rows where AEROPUERTO_ESCALA_CODE hasnt matched (deleted territories like Total)
df_f.dropna(inplace=True)

df_f["AEROPUERTO_ESCALA_CODE"] = df_f["AEROPUERTO_ESCALA_CODE"].astype(int)

df_f["SERVICIO_AEREO_CODE"] = df_f["SERVICIO_AEREO_CODE"].map(dict(zip(airservice['AirServiceCode'], airservice['AirServiceId'])))

df_f["MOVIMIENTO_AERONAVE_CODE"] = df_f["MOVIMIENTO_AERONAVE_CODE"].map(dict(zip(aircraftmovement['AircraftMovementCode'], aircraftmovement['AircraftMovementId'])))
```

```python
df_f['TIME_PERIOD_CODE'] = pd.to_datetime(df_f['TIME_PERIOD_CODE'].str.replace('M', ''), format='%Y-%m')
```
